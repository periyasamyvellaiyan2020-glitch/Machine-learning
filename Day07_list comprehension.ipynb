Tricky Problem: Pythagorean Triplets
You are given an integer N.
Print a list of all triplets [a, b, c] where:
1 ≤ a ≤ b ≤ c ≤ N
a² + b² = c²
Use a single list comprehension (no separate loops).
Each triplet should appear only once, so [3, 4, 5] is valid but [4, 3, 5] must not appear.

```python
n=int(input())
print([[a,b,c] for a in range(1,n+1)  for b in range(1,n+1) for c in range(1,n+1) if a<b<c if((a*a) + (b*b))==(c*c)])
```
