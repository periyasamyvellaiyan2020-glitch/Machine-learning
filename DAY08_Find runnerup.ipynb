Given the participants' score sheet for your University Sports Day, you are required to find the runner-up
score. You are given n scores. Store them in a list and find the score of the runner-up.

Input Format

The first line contains n. The second line contains an array A[] of n integers each separated by a

space.

Constraints

. 2≤n≤ 10

. -100 ≤ A[i] ≤100

Output Format

Print the runner-up score.

Sample Input O

5
23665

Sample Output O

5

```python
import numpy as np
```

```python
n = int(input())
arr = list(map(int, input().split()))   #map will return map object it wont give output
maximum=arr[0]
second_maximum=arr[1]
# print(arr)
for i in arr[2:]:
    if maximum < i:
        second_maximum=maximum
        maximum=i             #finiding maximum
    elif second_maximum<i:
        second_maximum=i       #second maximum
print(second_maximum)
    
```
