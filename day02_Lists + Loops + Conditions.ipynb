You have this list of customer ages:
ages = [22, 35, 41, 28, 19, 52, 33, 45, 27, 61, 38, 24, 55, 31, 42]


Your task
Write a Python program that:
1. Finds the average age.
2. Finds how many customers are above the average age.
3. Finds how many customers are below the average age.
4. Creates a new list containing only the ages above the average.
5. Finds the youngest and oldest age.
Expected type of output
Average age: ...
Above average count: ...
Below average count: ...
Above average ages: [...]
Youngest age: ...
Oldest age: ...

⚠️ Rules
Try to solve it using only the Python concepts you've already learned:
- for loop
- if
- list
- variables
- len()
- basic arithmetic
Don't use NumPy or Pandas.

```python
ages = [22, 35, 41, 28, 19, 52, 33, 45, 27, 61, 38, 24, 55, 31, 42]
# Finds the average age.
count=0
total=0
for i in ages:
    count+=1
    total+=i
avg=total/count
print('Average age:',f"{avg:.2f}") #36.86

# Finds how many customers are above the average age.
above_avg_count=0
above_avg=[]
for i in ages:
    if i>avg:
        above_avg_count+=1
        above_avg.append(i)
print('above the average age count:',above_avg_count)  
print('Above average ages',above_avg)

below_avg_count=0
below_avg=[]
for i in ages:
    if i<avg:
        below_avg_count+=1
        below_avg.append(i)
print('below the average age:',below_avg_count)  
print('Below average ages',below_avg)

Youngest_age=ages[0]
for i in ages[1:]:
    if Youngest_age>i:
        Youngest_age=i
print('Youngest age:',Youngest_age)
oldest_age=ages[0]
for i in ages[1:]:
    if oldest_age<i:
        oldest_age=i
print('Oldest age:',oldest_age)



```

shorter form


```python
ages = [22, 35, 41, 28, 19, 52, 33, 45, 27, 61, 38, 24, 55, 31, 42]

avg = sum(ages) / len(ages)

above_avg = []
below_avg = []

for age in ages:
    if age > avg:
        above_avg.append(age)
    elif age < avg:
        below_avg.append(age)

print("Average age:", f"{avg:.2f}")
print("Above average count:", len(above_avg))
print("Above average ages:", above_avg)
print("Below average count:", len(below_avg))
print("Below average ages:", below_avg)
print("Youngest age:", min(ages))
print("Oldest age:", max(ages))
```
