```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "age":          [45, 52, np.nan, 61, 38, 49, 55, np.nan, 42, 58],
    "bp":           [120, 118, 125, 135, np.nan, 122, 128, 119, np.nan, 130],
    "tumor_volume": [12, 14, 15, np.nan, 13, 150, 16, 14, np.nan, 17],
    "smoker":       ["yes", "no", "no", "yes", "no", np.nan, "no", np.nan, "yes", "no"],
    "tumor_site":   ["Lung", "Breast", "Lung", "Lung", np.nan, "Breast", "Lung", np.nan, "Lung", "Breast"],
})

train = df.iloc[:7].copy()      # first 7 rows
test  = df.iloc[7:].copy()      # last 3 rows

number=['age','bp','tumor_volume']
text=['smoker','tumor_site']

number_imp=SimpleImputer(strategy='mean')
text_imp=SimpleImputer(strategy='most_frequent')

train[number]=number_imp.fit_transform(train[number]).round(2)
train[text]=text_imp.fit_transform(train[text])

test[number]=number_imp.transform(test[number]).round(2)
test[text]=text_imp.transform(test[text])
test
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(2026)
n = 50

df = pd.DataFrame({
    "age":              rng.normal(38, 9, n).round(0).clip(21, 65),
    "years_experience": rng.normal(12, 6, n).round(0).clip(0, 40),
    "monthly_income":   rng.lognormal(10.7, 0.5, n).round(0),
    "credit_score":     rng.normal(700, 55, n).round(0).clip(300, 900),
    "loan_amount":      rng.lognormal(6.0, 0.8, n).round(0),
    "num_dependents":   rng.integers(0, 5, n).astype(float),
    "education":        rng.choice(["High School", "Bachelor", "Master", "PhD"], n, p=[0.2, 0.45, 0.28, 0.07]),
    "city":             rng.choice(["Chennai", "Coimbatore", "Bengaluru", "Hyderabad"], n),
    "owns_home":        rng.choice(["yes", "no"], n, p=[0.55, 0.45]),
    "defaulted":        rng.integers(0, 2, n),
})

holes = {"age": 4, "years_experience": 5, "monthly_income": 6, "credit_score": 5,
         "loan_amount": 4, "num_dependents": 3, "education": 4, "owns_home": 3}
for col, k in holes.items():
    df.loc[rng.choice(n, k, replace=False), col] = np.nan

df.loc[rng.choice(n, 4, replace=False), "city"] = "unknown"

df.head()   # Here I'm checking what are the columns or belongs to Text out of the columns of numerical in that I am categorising into which columns can be used as a mean median And mode

train=df[0:30]  # Here I am splitting the data 
validation=df[31:41]
test=df[42:50]

mean_list=['age','years_experience','defaulted']
median_list=['monthly_income','credit_score','loan_amount','num_dependents']
text_list=['education','city','owns_home']

train['city']=train['city'].replace("unknown",np.nan)  # Hangling this unknown values I'm converting to the missing value NA

mean_list_imp=SimpleImputer(strategy='mean')
median_list_imp=SimpleImputer(strategy='median')
text_list_imp=SimpleImputer(strategy='most_frequent')

train[mean_list]=mean_list_imp.fit_transform(train[mean_list])  # for missing values Finding the Median mean From the training data And then filling in Training data
train[median_list]=median_list_imp.fit_transform(train[median_list])
train[text_list]=text_list_imp.fit_transform(train[text_list])

validation[mean_list]=mean_list_imp.transform(validation[mean_list])    # For missing values In the validation data Fillling the training Data values
validation[median_list]=median_list_imp.transform(validation[median_list])
validation[text_list]=text_list_imp.transform(validation[text_list])
 
test[mean_list]=mean_list_imp.transform(test[mean_list])   # misssing values in the testing data ,fillingthe values from Training data
test[median_list]=median_list_imp.transform(test[median_list])
test[text_list]=text_list_imp.transform(test[text_list])


```
