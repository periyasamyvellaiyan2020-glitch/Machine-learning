<!-- Write a Python program to find:
1. Number of positive values
2. Number of negative values
3. Number of zeros
Expected output
Positive numbers: 4
Negative numbers: 3
Zeros: 2 -->

<!-- Rules 🎯
Use only:
- for loop
- if / elif / else
- variables
- len() if you need it
❌ Don't use:
- count()
- NumPy
- Pandas
- Counter
- external libraries -->

```python
numbers = [1000000, -999999, 0, 500000, -1] #This is the given list

positive=[]
negative=[]
zeros=[]
for i in numbers:
    if i<0:                                     # Here I am filtering out negative numbers 
        negative.append(i)
    elif i==0:                                  # Here I'm filtering out zeros 
        zeros.append(i)
    else:                                          # Here I am filtering positive numbers 
        positive.append(i)
print(' Number of positive values|:',len(positive))
print(' Number of negative values|:',len(negative))
print(' Number of zeros values|:',len(zeros))
```
