```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
```

```python
x,y=make_moons(n_samples=22000,noise=0.25,random_state=42)
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

base_model = LogisticRegression(random_state=42)
base_model.fit(X_train, y_train)
base_model_train=base_model.predict(X_train)
base_model_test=base_model.predict(X_test)
base_train_acc=accuracy_score(y_train, base_model.predict(X_train)) * 100
base_test_acc = accuracy_score(y_test, base_model.predict(X_test)) * 100
print(f"1.BASE MODEL TRAIN ACCURACY {base_train_acc:.2f} %")
print(f"1. Base Model (Logistic Regression) Test Accuracy: {base_test_acc:.2f}%")


xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)
xgb_model.fit(X_train, y_train)

xgb_train=xgb_model.predict(X_train)
xgb_test=xgb_model.predict(X_test)

xgb_train_score=accuracy_score(y_train,xgb_train)*100
xgb_test_score=accuracy_score(y_test,xgb_test)*100

print(f" XG booster train score {xgb_train_score:.2f} %")
print(f' XGB test score {xgb_test_score:.2f} %')
```
