```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(7)
n = 80

cities = ['Chennai', 'Coimbatore', 'Madurai', 'Trichy',
          'Salem', 'Erode', 'Vellore', 'Kanyakumari']
p = [0.28, 0.24, 0.18, 0.12, 0.08, 0.05, 0.03, 0.02]

df = pd.DataFrame({
    'age': rng.integers(25, 80, n),
    'gender': rng.choice(['Male', 'Female'], n),
    'city': rng.choice(cities, n, p=p),
    'treatment': rng.choice(
        ['Radiotherapy', 'Chemotherapy', 'Surgery', 'Immunotherapy'],
        n, p=[0.4, 0.3, 0.2, 0.1]),
    'stay_days': rng.integers(1, 15, n),
})
df['readmitted'] = rng.integers(0, 2, n)

X = df.drop(columns='readmitted')
y = df['readmitted']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=32
)
```

```python
bank=pd.read_csv(r'C:\Users\VP\Downloads\bank.csv')
from sklearn.model_selection import train_test_split  
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

y=bank['deposit']
x=bank.drop(columns='deposit')
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)

text=x_train.select_dtypes(include='str').columns
number=x_train.select_dtypes(include=['int','float']).columns

si=SimpleImputer(strategy='most_frequent')
for i in text:
    x_test[i]=x_test[i].replace('unknown',np.nan)
    x_train[i]=x_train[i].replace('unknown',np.nan)


x_train[text]=si.fit_transform(x_train[text])
x_test[text]=si.transform(x_test[text])

```

```python
ohe=OneHotEncoder(handle_unknown='ignore', sparse_output=False)
train_array=ohe.fit_transform(x_train[text])
train_ohe = pd.DataFrame(train_array,
                         columns=ohe.get_feature_names_out(text),
                         index=x_train.index)
```

```python
x_train_final=pd.concat([train_ohe,x_train[number]],axis=1)
```
